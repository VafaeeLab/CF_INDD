```python
df_nv = pd.read_excel("Analysis-Case_6Y.xlsx")
df_nv = df_nv.set_index('Flow Rate', drop=False)

df_nv = df_nv[df_nv['Location'] == 'Nasal Valve']
df_nv =df_nv[['Class1','Class2','Class3','Class4', 'Class5','Class6', 'Class7','Class8', 'Class9', 'Class10']]

plt.figure(figsize=(12, 5))

sns.heatmap(df_nv,cmap="jet", cbar_kws={'label': 'Percent'})
plt.ylabel('Flow rate [L/min]', fontsize = 18,labelpad = 10)
plt.xlabel('Diameter Class', fontsize= 18)
plt.title('Nasal Valve',fontsize=20, fontweight = 'bold')
plt.show()
```
